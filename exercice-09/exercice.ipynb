```python
import pandas as pd
titanic = pd.read_csv("../exercice-08/data/titanic_propre.csv")
print(titanic.shape)
```

```python
taux_survies = titanic.groupby(["pclass","sex"])["survived"].mean()
print(taux_survies)
```

**Réponse 1 :** Oui, les femmes ont mieux survécu que les hommes dans chacune des trois classes. En 1re classe, 97 % des femmes ont survécu contre 37 % des hommes ; en 2e classe, 92 % contre 16 % ; en 3e classe, 50 % contre 13,5 %.
**Réponse 2 :** Oui, en partie. Les enfants ont le taux de survie le plus élevé (58 %), contre 48 % pour les ados, 36 % pour les adultes et 27 % pour les seniors : plus on est jeune, plus on a survécu. Mais 42 % des enfants n'ont pas survécu, donc ils ont été favorisés sans être tous sauvés.
**Réponse 3 :** Cherbourg a le prix moyen le plus élevé (environ 60), devant Southampton (27) et Queenstown (13). La répartition des classes explique une partie de l'écart : 51 % des passagers de Cherbourg voyagent en 1re classe, contre 20 % à Southampton et 3 % à Queenstown. Mais elle n'explique pas tout : à classe égale, Cherbourg reste plus cher en 1re classe (105 contre 70 à Southampton) et en 2e classe (25 contre 20), alors que c'est l'inverse en 3e classe (11 contre 15). Pour Queenstown, trop peu de passagers en 1re et 2e classe pour comparer.
**Réponse 4 :** Voyager seul diminuait les chances de survie : 30 % des passagers seuls ont survécu, contre 51 % de ceux qui voyageaient avec de la famille. Cet écart ne prouve pas que voyager seul en soit la cause : les passagers seuls étaient aussi plus souvent des hommes et des passagers de 3e classe, deux groupes qui survivaient moins.

```python
titanic["tranche_age"] = pd.cut(titanic["age"], bins=[0, 12, 17, 59, 100], labels=["enfant", "ado", "adulte", "senior"]) 
print(titanic["tranche_age"].value_counts()) 
taux_survies_tranche = titanic.groupby("tranche_age")["survived"].mean()
print(taux_survies_tranche)
```
