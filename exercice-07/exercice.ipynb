```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
```

```python
passagers_mineurs=titanic[titanic["age"]<18]
nb_moins_18 = len(passagers_mineurs)
print(nb_moins_18)
```

```python
titanic_trie = titanic.sort_values("fare", ascending=False)
cinq_plus_chers = titanic_trie.head(5)
billets_chers = cinq_plus_chers[["fare", "class", "embark_town"]]
print(billets_chers)
```

```python
premiere_classe = titanic[titanic["pclass"] == 1]                         
nb_premiere_classe = len(premiere_classe)                                        

femmes_premiere_classe = premiere_classe[premiere_classe["sex"] == "female"]        
nb_femmes_premiere_classe = len(femmes_premiere_classe)                             

part_femmes_premiere_classe = nb_femmes_premiere_classe / nb_premiere_classe * 100   
print(f"Femmes en 1re classe : {part_femmes_premiere_classe:.1f} %")                 
```
