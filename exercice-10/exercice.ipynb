```python
import pandas as pd
titanic = pd.read_csv("../exercice-08/data/titanic_propre.csv")
print(titanic.shape)
```

```python
ports = pd.DataFrame({
    "embark_town":["Southampton", "Cherbourg", "Queenstown"],
    "pays":["Angleterre","France","Irlande"],
})
print(ports)
```

```python
nb_avant = len(titanic)
titanic = titanic.merge(ports, on= "embark_town", how="left" ,validate="many_to_one")
nb_apres = len(titanic)
print(nb_avant , nb_apres)
assert nb_avant == nb_apres 
```

```python
nb_passagers_pays= titanic["pays"].value_counts()
print(nb_passagers_pays)
nb_survivants_pays = titanic.groupby("pays")["survived"].mean()
print(nb_survivants_pays)
```

```python
ports_sans_q = ports[ports["embark_town"] !=  "Queenstown"]
print(ports_sans_q)
base = titanic.drop(columns=["pays"])
jointure_inner = base.merge(ports_sans_q , on = "embark_town", how = "inner")
jointure_left = base.merge(ports_sans_q  , on = "embark_town", how = "left")
print(len(jointure_inner), len(jointure_left))
```

```python
nb_avant = len(titanic)
titanic = titanic.merge(ports, on= "embark_town", how="left" ,validate="many_to_one")
nb_apres = len(titanic)
print(nb_avant , nb_apres)
assert nb_avant == nb_apres 
```

**Retirer Queenstown de `ports` :** avec `how="inner"`, on ne garde que les passagers dont le port est dans les deux tables : il reste 812 lignes, car les 77 passagers de Queenstown et les 2 passagers "Inconnu" disparaissent. Avec `how="left"`, on garde les 891 passagers, mais ceux de Queenstown et "Inconnu" ont un pays vide (`NaN`). Pour ne perdre aucun passager, `left` est le plus sûr, à condition de vérifier ensuite les pays vides.
